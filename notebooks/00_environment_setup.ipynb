```python

from pathlib import Path

# Main project directory
PROJECT_ROOT = Path(
    "/content/drive/MyDrive/YOLOv8_Fault_Tolerance"
)

# Code repository directory
CODE_ROOT = PROJECT_ROOT / "code"

# Create the project folders
folders = [
    "configs",
    "models",
    "data",
    "losses",
    "training",
    "fault_injection",
    "fault_tolerance",
    "experiments",
    "evaluation",
    "notebooks",
    "results",
    "tests",
]

PROJECT_ROOT.mkdir(parents=True, exist_ok=True)
(CODE_ROOT / "notebooks").mkdir(parents=True, exist_ok=True)

for folder in folders:
    (CODE_ROOT / folder).mkdir(parents=True, exist_ok=True)

# Create initial project files without overwriting existing files
starter_files = {
    "README.md": """# YOLOv8 Fault Tolerance

Research project on fault injection and fault tolerance
in a modular YOLOv8n object detection model.

## Project status
Phase 1: Environment setup and project scaffolding.
""",
    "requirements.txt": """# Initial dependencies
# Exact versions will be pinned after environment validation.
numpy
matplotlib
pyyaml
pytest
""",
    ".gitignore": """# Python
__pycache__/
*.py[cod]
.ipynb_checkpoints/
.pytest_cache/
.venv/
venv/

# Secrets
.env
*.key

# Model weights and large artifacts
*.pt
*.pth
*.onnx
*.engine

# Experiment outputs
artifacts/
datasets/
results/
runs/
wandb/

# OS files
.DS_Store
""",
}

for filename, content in starter_files.items():
    path = CODE_ROOT / filename

    if not path.exists():
        path.write_text(content, encoding="utf-8")
        print("Created:", path.relative_to(CODE_ROOT))
    else:
        print("Already exists, preserved:", path.relative_to(CODE_ROOT))

# Display the created structure
print("\nProject structure:\n")
for path in sorted(CODE_ROOT.rglob("*")):
    if path.is_dir():
        print("📁", path.relative_to(CODE_ROOT))
    elif path.name in starter_files:
        print("📄", path.relative_to(CODE_ROOT))

print("\nProject root:", PROJECT_ROOT)
print("Code root:", CODE_ROOT)
```
