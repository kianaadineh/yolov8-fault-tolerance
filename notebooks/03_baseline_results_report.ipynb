# YOLOv8s — Official COCO2017 Baseline Results
This notebook reads previously saved official COCOeval metrics.

**No inference or GPU execution is required.**

```python
from pathlib import Path
import json
from google.colab import drive

drive.mount('/content/drive')

metrics_path = Path(
    '/content/drive/MyDrive/YOLOv8_Fault_Tolerance/code/'
    'results/official_cocoeval/'
    'yolov8s_official_cocoeval_metrics.json'
)

assert metrics_path.is_file(), 'Saved metrics file not found'
record = json.loads(metrics_path.read_text(encoding='utf-8'))
print('Metrics file loaded successfully')
```

```python
metrics = record['cocoeval']['metrics']

assert record['num_images'] == 5000
assert record['num_predictions'] == 570662
assert len(metrics) == 12

print('Official YOLOv8s COCO2017 results')
for name, value in metrics.items():
    print(f'{name}: {value:.6f}')
```
